```python
import pandas as pd

import scraper
import exam_dataframe
import exam_modifiers
import pdf_scraper
```

```python
soup, buttons = scraper.scrape()
```

```python
bachelor_module = pdf_scraper.extract_modulenumbers()
```

```python
master_module = pdf_scraper.extract_modulenumbers(bachelor=False)
```

```python
module_gesamt = pdf_scraper.concatenate_bachelor_and_master(bachelor_module, master_module)
```

```python
klausurdaten = exam_dataframe.build_dataframe(soup, buttons)
```

```python
klausurdaten_replaced = exam_modifiers.replace_semester(klausurdaten)
```

```python
klausurdaten_replaced_time = exam_modifiers.time_proxy(klausurdaten_replaced)
```

```python
klausurdaten_replaced_time_summarized = exam_modifiers.summarize_vor_nachklausur(klausurdaten_replaced_time)
```

```python
klausurdaten_replaced_time_summarized_expanded = exam_modifiers.expand_wintersemester(klausurdaten_replaced_time_summarized)
```

```python
klausurdaten_replaced_time_summarized_expanded_avg = exam_modifiers.berechne_durchschnittsnote(klausurdaten_replaced_time_summarized_expanded)
```

```python
klausurdaten_replaced_time_summarized_expanded_avg_studiengang = exam_modifiers.fuege_studiengang_hinzu(klausurdaten_replaced_time_summarized_expanded_avg
                                                                                                , module_gesamt)
```

```python
klausurdaten_replaced_time_summarized_expanded_avg_studiengang_conc = exam_modifiers.concatenate_module(klausurdaten_replaced_time_summarized_expanded_avg_studiengang)
```

```python
klausurdaten_replaced_time_summarized_expanded_avg_studiengang_conc.to_csv("../data/klausurdaten.csv", index=False)
```
